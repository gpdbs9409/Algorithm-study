가장 적은 비용부터
이미 연결된 섬의 그룹에 새 섬을 추가 
    한쪽 섬이 그룹에 있으면
    다른쪽도 추가

#알고리즘: 최소비용 = 그리디 , 선으로 연결 = 그래프

```python
def solution(n, costs):
    answer = 0
    costs.sort(key = lambda x:x[2])
    connect = set([costs[0][0]]) #첫번째 섬, 어디를 연결할껀지 뻗는
    #지금까지 어디와 연결되었는지를 추적하기 위해
    while len(connect) !=n: #연결된 섬의 개수가 전체 개수와 동일할때까지
        for cost in costs:
            if cost[0] in connect and cost[1] in connect:
                continue
            elif cost[0] in connect or cost[1] in connect:
                connect.update([cost[0],cost[1]])
                answer += cost[2]
                break
    return answer
```
