##알고리즘 선택 이유
1. 가장 무거운 사람 탈때 가장 가벼운 사람을 타야 '최적'이기 때문에 -> 그리디
2. 무거운, 가벼운 짝지어서 pop 해야하기 떄문에 오름차순으로 정렬해서 양 끝을 뺴는 deque를 사용해야 겠다고 판단

#일단 사람을 오름차순 정렬하고
#배열을 덱으로 바꾸고
# [-1]이 limit보다 크다면, pop하고 구명보트 개수를 +1 한다
#그렇지 않담면, [-1]과 [0]이 묶어서, 둘의 합이 limit보다 아래라면 [-1]을 pop하고 다시 구면보트 개수 +1 바복
#배열이 Empty가 될때까지 계속 반복

```python
from collections import deque

def solution(people, limit):
    people.sort()
    people = deque(people)

    boat = 0
    while people:
        if len(people) == 1 or people[-1] + people[0] > limit:
                    people.pop()
                    boat+=1
        else: 
                    people.pop()
                    people.popleft()
                    boat+=1

    return boat
```
