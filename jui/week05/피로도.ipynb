#### 왜 이 알고리즘?
백트래킹: 재귀적으로 문제를 해결, 조건에 위배되는지 판단, 위배되면 해당상태를 제외하고 다음단계로 넘어가는 것, 경우의 수에서 조거능ㄹ 만족하는 것 -> 완전탐색기법 bfs, dfs로 구현이 가능함 dfs가 더 가까운
#### 시간 복잡도
O(N x N!)
던전의 개수 N
순열 탐색 필요 , N개의 순서를 정하는 함수 N!
최대 8개 => 8! = 40320번
완전탐색 가능 

#### 공간 복잡도
O(N^2)
재귀 + 슬라이싱

#### 엣지 케이스
1. 던전 탐색 불가능
초기 count = 0 반환
2. 피로도 k가 크다 = N개수 반환 
3. 동일 수치 단전 = 중복 데이터도 정상 탐색
4=

#### 새로 배운 것
1. 파이썬 함수 호출 안에서는 -= 쓸 수없음! 
2. 상태값 : 부모에서 자식으로 넘어가는 값과, 
결과값: 해당 분기에서 얻어낸 결과값을 확실히 분리하기

```python
def solution(k, dungeons):
    #던전의 개수는 8 이하
    #최소 필요도 >= 소모 필요도 
    # 서로 다른 던전의 [최소 필요 피로도, 소모 피로도] 같을 수 있다

    #dangeons 각 던전의 최소 필요도, 소모 필요도
    
    def dfs(k,dungeons, count):
        max_count = count #부모에서 자식으로 넘어가는 상태값
        for i in range(len(dungeons)):
            if dungeons[i][0] <=k: 
                new_dungeops = dungeons[:i] + dungeons[i+1:]
                max_count =  max(max_count,dfs(k - dungeons[i][1] ,new_dungeops, count+1))  #이렇게 하면 i부터 할 수있는데 k값이 그림 계속 초기화 되야하는거 아닌가?
                #count는 현재 이 재귀 깊이까지 오면서 실제로 탐험한 던전 수

                #for 문담 바퀴를 돌떄, 이전 count 결과가 넘어간다

    #최대한 많이 탐험하려 한다
        return max_count
    return dfs(k,dungeons,0)
```

```python
#이전 풀이

def solution(k, dungeons):
    answer = 0
    visited = [False] * len(dungeons)
    
    def dfs(fatigue, count):
        nonlocal answer
        answer = max(answer, count)
        
        for i in range(len(dungeons)):
            need, use = dungeons[i]
            
            if not visited[i] and fatigue >= need:
                visited[i] = True
                dfs(fatigue - use, count + 1)
                visited[i] = False
    
    dfs(k, 0)
    return answer
```
