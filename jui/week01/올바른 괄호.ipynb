가장 최근 가장 먼저 처리 => 스택이라고 판단

4. 새롭게 배운 내용
스택 비어있는건 not stack으로 짧게 표현 가능


```python
def solution(c):
    stack = []
    
    for i in c:
        if i == '(':
            stack.append('(')
        else:  # i == ')'
            if not stack:      # 스택이 비어있다면
                return False
            stack.pop()
    
    # 반복문 끝난 뒤
    if stack:   # 스택이 남아있다면
        return False
    else:
        return True
```
