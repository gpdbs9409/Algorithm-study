k 개의 수를 제거
numberdㅔ서 k개를 제거했을때 가장 커야 하니까
number에서 가장 큰 수 기준으로 그 앞에 있는 수 중에 가장 작은 수 하나씩 삭제하면 
만약 앞에걸 다 제거했는데도 k가 더 있다면, 뒤에 있는 수중에 작은 수 순서대로 지워간다

#새롭게 알게된 것
1. max() 시간복잡도 O(N), 최소한으로 쓰기
2. pop에는 값이 아닌 인덱스를 넘겨줘야 한다

```python
#시간복잡도 O(n2)

def solution(number, k):

    #number 에 있는 수들을 하하나씩 배열로 만들고
    #리스트 전체는 대괄호로 감싸야 한다
    number = [i for i in number]
    count = k

    while(count):
        #가장 큰 수의 인덱스 찾아서
        m = number.index(max(number))

        if m > 0:

            #그보다 작은 범위 내의 작은수를 찾음
            n = number.index(min(number[0:m]))
            number.pop(n)
            count-=1

        #그리고 제거해야 할 수가 제일 큰 수의 앞에 있는 개수보다 많다면
        else:
                    s = number.index(min(number))
                    number.pop(s)
                    count-=1

    return "".join(number)
```

스택에 값을 2개씩 넣는데,
만약 이전의 값보다 크다면 그것 유지 

앞의 자리의 수가 클수록 좋다 -> 점점 증가하는 모양세가 좋다

```python
def solution(number, k):
    stack = []
    
    for num in number:
        # 내 앞의 숫자가 나(num)보다 작고, 아직 지울 기회(k)가 남아있다면
        # 내 앞의 작은 숫자를 지워서 나(더 큰 수)를 앞으로 당긴다!
        while stack and k > 0 and stack[-1] < num:
            stack.pop()
            k -= 1
            
        stack.append(num)
        
    if k > 0:
        stack = stack[:-k]
        
    return "".join(stack)
```
