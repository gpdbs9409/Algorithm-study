최솟값 -> Greedy

#현재 상황에서 최적의 방법을 찾아야 하는데

#AA 부터 시작하니까, 전체 알파벳 갯수 36개 
#그 절반은 18 -> R! 

#만약 현재 커서가 +18, -18이랑 가까운지 계산해서 R 하던가 L 하던가 해야겠네

알파벳 변경 / 커서 이동(좌우) 분리해서 풀어야 함
좌우 = 왼쪽 오른쪽의 최소 횟수

알게된 것
1. 파이썬에서는 문자끼리 - 연산으로 뺄 수 없음 (아스키코드 불가) -> ord() 사용해야함

```python
def solution(name):
    count = 0
    n_len = len(name)
    
    lst = [False] * n_len
    idx = 0 #현재 커서 위치
    n = name[0]

    while lst[idx] == False:
        #가로로 쭉 보고, 
        #목표로하는글자까지 카운트하기 
        n = name[idx]
        count+=min(ord('Z') - ord(n)-1,ord(n) - ord('A'))
        lst[name.index(n)] = True

    while lst[idx] == False:
        r_idx = idx + 1
        if r_idx == n_len:
            r_idx = 0

        l_idx = idx - 1
        if l_idx == -1:
            l_idx = n_len-1

        if lst[l_idx] == False:
            idx = l_idx
        elif lst[r_idx] == False:
            idx = r_idx
        #그다음에 lst[idx] == False를 찾아가야하는데, 
        # 왼쪽으로 가는게 빠른지 오른쪽으로 가는게 빠른지 계산


    return count
```
