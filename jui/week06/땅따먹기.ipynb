### 왜 이 알고리즘?
1. 처음에는 현재 행의 최댓값, 이전 행에서 그 열을 선택했다면 차선을 고르기로 했다
그렇게 하면 전체 최대값 (이전 값이 차선 현재 값이 최선)인 경우를 구하기가 힘들어서

### 시간 복잡도
새로 10^5 이하
n개의 행까지 4개의 열을 비교 -> 정수배
O(N)

### 공간 복잡도
N행 x 4열
O(N)

### 엣지 케이스

### 새로 배운 것
1. dp = [[0][0][0][0]] * n
[0][0]은 0 번째 원소로 인식이 되서, 접근 오류
2. dp = [[0] * 4] * n
모든 행이 같은 리스트를 '공유'함
3. dp[0] = land[0]
두 변수가 같은 포인터를 가르켜서, dp[0]을 수정하면 land[0]도 바뀜
dp[0] = land[0][:] => 별도의 리스트로 복사! 

```python
def solution(land):
    #N행 4열
    # 가로로 땅을 밟으며 한 행씩 내려오는데
    # 4개의 가로 칸 중의 하나만 밟아야 한다 
    # 같은 열 반복해서 밟을 수 없음

    #4가지로 경우의 수가 나눠지는 트리

    #dp는 각각 열에 도달했을때 최대값 
    n = len(land)
    dp = [[0] * 4 for _ in range(n)]

    dp[0] = land[0]

    for i in range(1,n):
        for j in range(4):
            #점화식 이전 행에서 j를 제외한 최대값에 현재 값 더하기
            dp[i][j] = land[i][j] + max(
                dp[i-1][x] for x in range(4) if x!=j
                )

    # 얻을 수 있는 점수의 최대값 
    return max(dp[-1])
```
