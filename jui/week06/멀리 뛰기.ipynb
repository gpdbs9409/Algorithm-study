### 왜 이 알고리즘?
1. 백트래킹으로 풀려고 , 중복 순열 product() 라이브러리를 생각했지만, 
n이 커져서 가능한 자리 수 (1,2,1,2,,,)가 커질 수록 -> n이 10만 되도, 10자리가 될 수 있음
만들어지는 경우의 수가 2의 n승으로, 리스트에 보관할 수 있는 경우의 수가 많아진다
2. 각각 () 안에 가능한 칸 수 만큼 반복되는 경우의 수를 저장해놓고 사용하면 문제 풀이가 빨라진다 -> DP 활용

### 시간 복잡도
n은 2000 이하! 2 x `10^3
2부터 n까지 순회하기 때문에
O(n)

### 공간 복잡도
길이가 n+1인 배열이기때문에
O(n)

### 엣지 케이스
n=1, n=2
1234567을 초과하는 dp[n]


### 새로 배운 것
1. combinations_with_replacements()와 product() 이것의 경우의 수 초과 염두해두기



```python
def solution(n):
    #멀리뛰기에 사용될 칸의 수
    
    #각 인덱스 칸까지 도착하는 방법의 수
    dp = [0] * (n+1)

    dp[0] = 1
    dp[1] = 1

    for i in range(2,n+1):
        dp[i] = (dp[i-1] + dp[i-2]) % 1234567


    #dp를 재사용해서, 각각 경우의 수 구하기

    #n칸에 도달하는 방법의 가지 수 

    return dp[n]
```
