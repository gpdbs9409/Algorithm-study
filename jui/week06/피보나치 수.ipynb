
### 왜 이 알고리즘?
재귀방식으로 처리하면, 똑같은 계산을 반복하는 비효율성이 있기 때문에
그 값을 저장하는 방식으로 처리 = DP 

### 시간 복잡도
O(N)
2부터 n까지 반복문 한번 실행

### 공간 복잡도
O(N)
크기가 n+1인 배열 사용

### 엣지 케이스
1. n이 1인 경우
2. n이 10^5 경우 
3. 모듈러 연산 활요하지 않으면, 시간초과! 

### 새로 배운 것
1. 모듈러 연산 = 미리 나머지를 구하고 더하거나 곱해도, 원본 큰 값에서 더하거나 곱한 결과값의 나머지는 동일하다
 = 계산하기 전에 미리 나누기 해서 수 자체의 크기를 줄여도 도니다
n의 값이 커서, 피보나치 수를 그대로 계산하면 값이 너무 커진다


```python
def solution(n):

    dp = [n] * (n+1)

    #0부터 시작하는것 주의 
    dp[0] = 0
    dp[1] = 1
    dp[2] = 1

    for i in range(3,n+1):
        dp[i] = (dp[i-1] + dp[i-2]) % 1234567

    #2 이상의 n이 입력되었을때, n번째 피보나치 수를 1234567으로 나눈 나머지 mod를 반환하기

    return dp[n]
```
