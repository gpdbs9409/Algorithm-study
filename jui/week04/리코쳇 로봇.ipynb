#### 새로 배운 것
1. move도 같이 큐에 넣어주기
2. range(board[0]): # ❌ 에러. 값으로 넣어줘야 함
3. 끝까지 이동하는 조건이어도, 무한루프에 빠지지 않게 하기위해 visited 필요! 
4. 범위 계산하는 식에서 board[nx][ny]를 읽으려고 하면 IndexError가 발생합니다.
5. set은 add() 사용!

```python
from collections import deque

def solution(board):
    #격자모양 게임판 위에서 말을 움직이는 게임
    #시작위치 ~ 목표위치
    #. 빈공간 R로봇 처음 위치 
    dx = [-1,1,0,0]
    dy = [0,0,-1,1]

    n = len(board)
    m = len(board[0])
    q = deque()
    visited = set()

    for i in range(n):
        for j in range(m):
            if board[i][j] == 'R':
                q.append((i,j,0))
                visited.add((i,j))

    move = 0

    
    
    while q:
        x, y, move = q.popleft()

        if board[x][y] == 'G':
            return move
        for i in range(4):
            nx = x
            ny = y
            while True: #한칸씩 상하좌우 방문하면서, 부딛힐때까지 q에 저장하는 bfs
                nnx = nx + dx[i]
                nny = ny + dy[i]

                    #장애물에 부딪혔다!
                if not (0 <= nnx < n and 0 <= nny < m) or board[nnx][nny] == 'D':
                    #다음 큐에 넣고 다음 부딪힘 착기!
                    break 
                    
                    
                #안부딪혔다
                nx = nnx
                ny = nny

            if (nx, ny) not in visited:
                q.append((nx,ny,move+1))  
                visited.add((nx,ny))
            
    return -1
```
