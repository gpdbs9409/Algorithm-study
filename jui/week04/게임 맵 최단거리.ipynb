#### 왜 이 알고리즘?
BFS : 시작점에서 출발해, 거리 순서대로 넓게 퍼져나감
DFS는 깊이로 먼저 도착하더라도, 그것이 최단경로라는 보장이 없음
가중치가 없어서 다익트스트라 x
#### 시간 복잡도
O(N^2) 까지 가능

#### 공간 복잡도
O(N^2) 
큐에 최악의 경우에는 맵 전체 칸 크기만큼

#### 엣지 케이스
1. 시작하자마자 갇혀있는 경우
(0,0) 주변이 전부 0
2. 맵의 크기가 1x1

#### 새로 배운 것
1. 큐 자료구조, 현재 위치 이동거리 넣어서  while queue: 돌리기
2. 튜플로 넣으면 3개의 인자를 바로 pop 가능
3. append는 하나만 뽑기 때문에 () 튜플로 묶어줘야 함

#### 문제 정리
게임 두 팀으로 나눠서 진행
상대팀 진영 파괴! = 이김
상대팀 진영에 빨리 도착해야 함

1,1 위치
상대팀 5,5
검은색 = 막힌 길
동서남북 한칸씩

상대팀이 벽을 세워뒀다면, 도착하지 못할 수도! 
상대방 위치는 map의 가로 x 세로



```python
from collections import deque

def solution(maps):
    #게임 상태, 2차원 배열
    #n 행 m 열, 인덱스
    n = len(maps) 
    m = len(maps[0])

#큐 자료구조, 현재 위치 이동거리 넣어서  while queue: 돌리기
    queue = deque([(0,0,1)]) #이걸로 배열에 접근해야 하므로, 0으로 하기
    #튜플로 넣기
    dx = [-1, 1, 0, 0]
    dy = [0, 0, -1, 1]

    maps[0][0] = 0
    #상하좌우 이동 정의해아 하지 않나?
    while queue:
        #큐에서 현재 위치 꺼내기
        x,y,dist = queue.popleft()

        #만약 목표에 도달했다면!
        if x == n-1 and y == m-1:
            return dist

        #좌표를 하나씩 바꿔감
        for i in range(4):
            nx = x + dx[i]
            ny = y + dy[i]
        
            #만약 다음 값이 0이면, 못감
            if 0 <= nx < n and 0 <= ny < m:
                if maps[nx][ny] == 1:
                    maps[nx][ny] = 0
                    queue.append((nx,ny,dist+1))
                
    return -1
    #몇개의 칸을 지나서 상대편 진영에 도착했는지 
    #상대팀 도착하기위해 지나야 하는 칸의 최솟값
   
```

```python
#이전에 풀었던 풀이

from collections import deque

def solution(maps):
    n = len(maps)
    m = len(maps[0])
    if maps[0][0] == 0:
        return -1
    
    dx = (-1,1,0,0)
    dy = (0,0,-1,1)

    #현재 있는 좌표
    q = deque()
    q.append((0,0))

    while q:
        x,y = q.popleft()

    #한 칸씩 이동하면서 만약 1이면 이동하고 0으로    
        for i in range(4):
            nx = x + dx[i]
            ny = y + dy[i]

            if nx >=n or ny>=m or nx<0 or ny<0:
                continue

            if maps[nx][ny] == 1:
                maps[nx][ny] = maps[x][y]+1
                q.append((nx,ny))


        #만약 마지막 최종 좌표가 5,5에 도달하지 못했다면
    if maps[n-1][m-1] == 1:
        return -1
    
    return maps[n-1][m-1]
```
