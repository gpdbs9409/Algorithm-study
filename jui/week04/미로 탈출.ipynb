#### 왜 이 알고리즘?
최단 거리 보장: 모든 간선의 가중치가 동일하다
BFS
#### 시간 복잡도
O(N x M)
행 x 열
각 칸은 레버 상태 (0,1)에 따라 최대 2번씩 큐에 들어간다

#### 공간 복잡도
O(N x M)
3차원 방문 배열의 크기 = N x M x 2
큐에 들어갈 수 있는 최대 원소 수 = n x m (미로전체 칸 수)

#### 엣지 케이스
1. 레버/출구 도달시
= 큐가 비어 -1리턴
2. 레버를 거쳐서 출구로 가는 도중 S을 다시 지나치는 경우
= visited 배열을 따로 관리하므로, 상관없음
3. 최소 격자 크기 
경게 조건 검사 따로 마련

#### 새로 배운 것
1. 레버를 당겼는가? 를 visited 배열 차원에 추가해서 관리
2. 레버를 당기기 전, 지나간 길이여도 당긴후 나가야 출구로 갈 수 있기때문에
, 단일 2차원 배열로 방문만 확인하면, 길이 막힐 수 있어 3차원으로 처리
3. 배열 범위 체크 함수를 먼저 배치! 


#### 문제 분석
너비우선탐색 = 최단 경로를 보장

```python
from collections import deque

#이동 가능한가?
def is_valid_move(ny,nx,n,m,maps):
 #범위 안인가
 #벽이 아닌가   
    return 0 <= ny < n and 0 <= nx < m and maps[ny][nx] != "X"
#true, false 반환

#방문한 적이 없으면 큐에 넣고 방문 여부 표시
def append_to_queue(ny,nx,k,time,visited,q):
    if not visited[ny][nx][k]:
        visited[ny][nx][k] = True
        q.append((ny,nx,k,time+1))

#k  = 레버 당겼는지 안당겼는지

def solution(maps):
    #n이 세로
    n, m = len(maps), len(maps[0])
    visited = [[[False]*2 for _ in range(m)] for _ in range(n)]
        
    q = deque()
    dy = [-1,1,0,0]
    dx = [0,0,-1,1]
    end_x = 0, end_y = 0

    #시작점과 도착점 찾아 큐에 넣고 방문 여부 표시
    for i in range(n):
        for j in range(m):
            if maps[y][x] == "S":
                q.append(i,j,0,0)
                visited[i][j][0] = True
            if maps[y][x] == "E":
                end_y, end_x = i,j #도착점

                            
    #큐에서 현재 위치 꺼낸뒤, for문 돌면서 4방향으로 가기
    while q:
        y,x,k,time = q.popleft()

        #출구 도착했다면?
        if y == end_y and x == end_x and k == 1:
            return time

        #이동을 하고
        for i in range(4):
            ny, nx = y + dy[i], x + dx[i]

            #만약 갈 수 있다면?
            if not is_valid_move(ny, nx,n,m,maps):
                continue

            #이동한 뒤에 레버 도달했다면
            if maps[ny][nx] == "L":
                append_to_queue(ny,nx,1,time,visited,q)
            else:
                append_to_queue(ny,nx,k,time,visited,q)

    return -1
        
    #간 길인지 체크하기 

        #S부터 L까지 최단거리 


    #L = 레버
    #레버를 먼저 당겨야 함!  -> 기록해야함

        #L부터 E까지 최단거리
    #E = 출구
    #만약 출구면, 종료하고 return 하기
    #만약 레버가 True가 아니면, 다시 돌아가야 함
    
    #ㅇ


    #미로 빠져나가는데 걸리는 시간
    #탈출 불가하면 -1
```
