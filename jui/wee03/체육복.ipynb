#앞뒤로 체육복을 빌려줄 수 있다

#1부터 n번까지의 학생이 있다

#이거 딕셔너리로 만들어서
#value값을 다 1로 채우고

#이중에서 체육복 가져온 사람들 
#만약 all_student의 key값이 reverse에 있으면 
#value 값을 2로 채우고

#만약 key값이 lost에 있으면
#value값을 0으로

#만약 all_student의 key값이 2인데 그 앞칸이 0이면, 앞칸을 1로 본값을 1로
#앞은 1인데 뒤가 0이라면, 뒤를 0으로

##알고리즘 
Greedy =>매 순간 최선의 선택(왼쪽 먼저 나눠주고 채워져 있으면 오른쪽 선택)
시간복잡도 O(n)

##배운점
1. 문제 예외조건 꼼꼼하게 읽어야 할 것
-> 처음에 풀때, lost 도난당하면 0이 되는 것으로 처리 
2. 중복 처리는 set()으로 


```python
def solution(n,lost,reverse):

    all_student = {i: 1 for i in range(1,n+1)}

    #중복 제거
    lost = list(set(lost))
    reverse = list(set(reverse))

    for r in reverse:
        if r in all_student:
            all_student[r] +=1 

    for l in lost:
            if l in all_student:
                all_student[l] -= 1

    for i in range(1,n+1): 
        if all_student[i] == 2:
         
            if i == 1:
                if all_student[2] == 0:
                    all_student[1] = 1  
                    all_student[2] = 1

            elif i == n:
                if all_student[n-1] == 0:
                    all_student[n] = 1  
                    all_student[n-1] = 1

            else: 
                if all_student[i-1] == 0:
                    all_student[i] = 1
                    all_student[i-1] = 1
                    
                
                elif all_student[i+1] == 0:
                    all_student[i] = 1
                    all_student[i+1] = 1

    answer = sum(1 for key in all_student if all_student[key] > 0)
    return answer
```

def solution(n, lost, reserve):
    all = [1] * n
    for i in lost:
        all[i-1] -= 1
    for i in reserve:
        all[i-1] += 1

        if all[i] == 0:
