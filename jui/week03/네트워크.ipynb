#### 왜 이 알고리즘?
dfs로 풀기 -> 선을 쭉 따라서 깊게 탐색하면 되는 전형적인 DFS 문제

#### 시간복잡도
O(N^2)
컴퓨터 개수 N일떄, N번 탐색하기 때문에
길이가 N인 배열 매번 순회

#### 공간복잡도
O(N)
visited 리스트 O(N)
재귀호출시 쌓이는 스택 O(N)


#### 엣지 케이스
1. 모든 컴퓨터가 하나로 연결된 경우
2. 개별 노드
3. 최대 입력값
4. 자기 자신과의 연결만 있는 경우 (computer[i][i]만 1)

#### 새롭게 알게된 것
1. dfs 함수 안에서 탐색 끝났다면 따로 체크하는 조건문 만들 필요 없음
2. count 자체는 dfs 밖에서 로직 구현해야 함 안,밖 구분 확실히


```python
def solution(n, computers):

    count = 0
    visited = [False] * n #노드 방문 확인
    def dfs(index):
        if visited[index] == True:
            return

        visited[index] =  True

        for j in range(n):
            if computers[index][j] == 1:
                dfs(j)

        

    
    for i in range(n):
        if visited[i] == False:
            dfs(i)
            count+=1
    return count
```
