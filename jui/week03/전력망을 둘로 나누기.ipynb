#### 왜 이 알고리즘?
트리 문제 : 간선 - 노드
작은 데이터 규모 
연결된 노드 세는 작업 : 전형적엔 DFS

#### 시간 복잡도
O(N^2)
간선 제거 N번 반복
DFS 시간복잡도 

#### 공간 복잡도
O(N)
visited 배열 O(N)
재귀호출 스택 O(N)

#### 엣지 케이스
1. 최소 입력 데이터 N=3
2. 스타 형태 트리 구조 (1번에 23456 다 연결)


#### 새로 배운 것
1. 빈 리스트에 요소 추가할 때는 .append
[i] = 1 할당 쓸 수 없음
2. N 개 , K-N개 (전체 - 부분) 잘 활용하기
3. 재귀함수 반환값 누적 += 활용
4. 리스트 슬라이싱 이용한 간선제거 -> pop()은 원본 데이터 훼손

```python
def solution(n, wires):
    
    #1번부터 시작!! 
    #해당 인덱스에서 연결되어있는 노드 개수가 몇개인지 반환해야 함  
    def dfs(idx,wires):
        idx_count = 1
        visited[idx] = True
        #현재 노드와 연결된 노드 찾기
        for a, b in wires:
            if a == idx:
                neighbor = b
            elif b == idx:
                neighbor = a
            else: continue

            #방문하지 않았다면 ,DFS 호출
            if visited[neighbor] == False:
                idx_count+=dfs(neighbor,wires)

        return idx_count
    
    answer = n
    #하나씩 끊어보는 코드
    for out in range(len(wires)):
        visited = [False] * (n + 1)
        new_wires = wires[:out] + wires[out+1:]

        minus = []
        #확인하는 코드
        count = 0
        for i in range(1,n+1):
            if visited[i] == False:
                minus.append(dfs(i,new_wires))
                count+=1
            
        if count == 2 :
            answer = min(answer,abs(minus[0] - minus[1]))
    return answer
```

```python
#이전에 풀었을 때 코드
def solution(n, wires):
    
    answer = n
    
    for i in range(len(wires)):
        
        graph = [[] for _ in range(n+1)]
        
        for j in range(len(wires)):
            if i != j:
                a, b = wires[j]
                graph[a].append(b)
                graph[b].append(a)
        
        
        visited = [False]*(n+1)
        
        
        def dfs(v):
            visited[v] = True
            count = 1
            
            for next in graph[v]:
                if not visited[next]:
                    count += dfs(next)
                    
            return count
        
        
        count = dfs(1)
        
        answer = min(answer, abs(count - (n-count)))
    
    
    return answer
```
