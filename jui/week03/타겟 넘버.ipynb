순서바꾸지 않고 - + 이용해서 타겟 넘버 만들기
numbers = 사용할 수 있는 숫자
target = 원하는 숫자

#알고리즘: DFS
선택 이유: 각 숫자마다 + - 선택 -> 2가지 가지뻗기 , 모든 경우의수 찾기

#시간복잡도
O(N2)

#공간복잡도
스택 깊이 = 재귀 깊이, 최대 n번 중첩 호출 
O(n) = 재귀 스택 깊이만 필요! 

```python
def solution(numbers , target):

    def dfs(index, current_sum):
    #지금 몇번쨰 숫자를 처리중?
    #current_sum =지금까지의 합

        #모든 숫자를 다 처리했을때
        if index == len(numbers):
            if  current_sum == target:
                    return 1
                #current_sum이 inex끝까지 값이니까, 
            else: 
                return 0
                 
        
        result = dfs(index + 1, current_sum + numbers[index])
        result += dfs(index + 1, current_sum - numbers[index])

        return result
    return dfs(0,0)
 
```
